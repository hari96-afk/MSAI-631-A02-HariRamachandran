```python
# ==============================
# 1. INSTALL REQUIRED LIBRARY
# ==============================
!pip install azure-ai-textanalytics

# ==============================
# 2. IMPORT LIBRARIES
# ==============================
import os
from azure.ai.textanalytics import TextAnalyticsClient
from azure.core.credentials import AzureKeyCredential

# ==============================
# 3. SET AZURE CREDENTIALS
# ==============================
os.environ["AZURE_ENDPOINT"] = "https://chatbot-language-service1.cognitiveservices.azure.com/"
os.environ["AZURE_KEY"] = "99vUQruCHRwwgBe7Je1QIZpAP06Oq6qlSrCkhgZNF4CtuHznIA74JQQJ99CEACYeBjFXJ3w3AAAaACOG9obo"

endpoint = os.environ["AZURE_ENDPOINT"]
key = os.environ["AZURE_KEY"]

# ==============================
# 4. CONNECT TO AZURE AI SERVICE
# ==============================
client = TextAnalyticsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(key)
)

# ==============================
# 5. SENTIMENT ANALYSIS FUNCTION
# ==============================
def analyze_sentiment(text):
    try:
        result = client.analyze_sentiment([text])[0]

        return {
            "text": text,
            "sentiment": result.sentiment,
            "positive": result.confidence_scores.positive,
            "neutral": result.confidence_scores.neutral,
            "negative": result.confidence_scores.negative
        }

    except Exception as e:
        return {"error": str(e)}

# ==============================
# 6. CHATBOT LOOP
# ==============================
print("🤖 Azure AI Chatbot (Sentiment Analysis)")
print("Type 'exit' to stop\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye 👋")
        break

    response = analyze_sentiment(user_input)

    if "error" in response:
        print("Bot Error:", response["error"])
    else:
        print("\nBot Analysis:")
        print("Sentiment:", response["sentiment"])
        print("Positive:", round(response["positive"], 3))
        print("Neutral:", round(response["neutral"], 3))
        print("Negative:", round(response["negative"], 3))
        print("-" * 40)
```
