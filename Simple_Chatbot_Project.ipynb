```python
print("Welcome to Student Support Chatbot")
print("Type 'bye' to exit")

while True:
    user = input("You: ").lower()

    if user == "hello":
        print("Bot: Hello! How can I help you?")

    elif user == "admission":
        print("Bot: Admissions are open until August.")

    elif user == "courses":
        print("Bot: We offer IT, Business, and Healthcare courses.")

    elif user == "fees":
        print("Bot: Please visit the college website for fee details.")

    elif user == "contact":
        print("Bot: You can contact us at support@college.edu")

    elif user == "bye":
        print("Bot: Goodbye!")
        break

    else:
        print("Bot: Sorry, I do not understand.")
```
