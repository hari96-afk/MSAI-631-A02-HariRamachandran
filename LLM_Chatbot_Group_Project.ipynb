```python
import gradio as gr
from transformers import pipeline
```

```python
chatbot_model = pipeline(
    "text-generation",
    model="distilGPT2"
)
```

```python
def generate_response(message, history):
    if message is None or message.strip() == "":
        return "Please enter a question."

    prompt = f"User: {message}/nAI:"

    response = chatbot_model(
        prompt,
        max_new_tokens=80,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
        pad_token_id=50256
    )

    generated_text = response[0]["generated_text"]
    answer = generated_text.replace(prompt, "").strip()

    if answer == "":
        answer = "I am not sure about that. Please try asking the question differently."

    return answer
```

```python
demo = gr.ChatInterface(
    fn=generate_response,
    title="LLM Chatbot for Human-Computer Interaction",
    description="Enter a question and the chatbot will generate a response using a Hugging Face model."
)

demo.launch(share=True)
```
