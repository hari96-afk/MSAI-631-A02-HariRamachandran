```python
import random

print("=" * 70)
print("AI-BASED ADAPTIVE QUIZ SYSTEM")
print("=" * 70)

questions = {
    "easy": [
        {"q": "What is 2 + 2?", "a": "4", "hint": "It is greater than 3"},
        {"q": "What color is the sky?", "a": "blue", "hint": "Same color as the ocean"}
    ],

    "medium": [
        {"q": "What is 12 x 3?", "a": "36", "hint": "12+12+12"},
        {"q": "Which planet do humans live on?", "a": "earth", "hint": "Third planet from the Sun"}
    ],

    "hard": [
        {"q": "What is the square root of 144?", "a": "12", "hint": "A number multiplied by itself gives 144"},
        {"q": "Who developed the Theory of Relativity?", "a": "einstein", "hint": "His first name is Albert"}
    ]
}

difficulty = "easy"

score = 0
correct_count = 0
wrong_count = 0

for question_number in range(1, 6):

    print("\n" + "-" * 70)
    print(f"Question {question_number}")
    print(f"Current Difficulty Level: {difficulty.upper()}")

    question = random.choice(questions[difficulty])

    answer = input(question["q"] + " : ").lower().strip()

    if answer == question["a"]:

        print("Result: CORRECT")

        score += 1
        correct_count += 1

        if difficulty == "easy":
            difficulty = "medium"
            print("Adaptive Action: Difficulty increased to MEDIUM")

        elif difficulty == "medium":
            difficulty = "hard"
            print("Adaptive Action: Difficulty increased to HARD")

    else:

        print("Result: INCORRECT")
        print("Correct Answer:", question["a"])

        wrong_count += 1

        print("AI Hint:", question["hint"])

        if difficulty == "hard":
            difficulty = "medium"
            print("Adaptive Action: Difficulty reduced to MEDIUM")

        elif difficulty == "medium":
            difficulty = "easy"
            print("Adaptive Action: Difficulty reduced to EASY")

accuracy = (score / 5) * 100

print("\n" + "=" * 70)
print("QUIZ COMPLETED")
print("=" * 70)

print("Final Score:", score, "/ 5")
print("Accuracy:", round(accuracy, 2), "%")

if accuracy >= 80:
    level = "Advanced User"
elif accuracy >= 50:
    level = "Intermediate User"
else:
    level = "Beginner User"

print("User Classification:", level)

print("\nAI Adaptive Summary")
print("--------------------")
print("Correct Answers:", correct_count)
print("Incorrect Answers:", wrong_count)

if correct_count > wrong_count:
    print("System Observation: User performs well with challenging questions.")
else:
    print("System Observation: User may benefit from easier questions and hints.")

print("=" * 70)
```
